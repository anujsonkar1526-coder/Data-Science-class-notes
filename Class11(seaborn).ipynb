```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("student_data.csv")
print(df)
```

```python
# line chat/plot

sns.lineplot(data=df, x='study_hours', y='exam_score')
```

```python
# Bar chart/plot

sns.barplot(data=df,x='grade',y='exam_score',color='red', estimatoe='sum')
```
