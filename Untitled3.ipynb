## :Total number of application in june (current month)

```python
a=df[df['issue_date'].dt.month==6] # given dataframe (this data only for june)

no_of_app=a['id'].nunique()

print(no_of_app)


# for current month


current_date=pd.Timestamp.today()   # timestampe:- given current date and time like (aaj wali date)
current_month= current_date.month
current_day = current_date.day
a=df[(df['issue_date'].dt.month==current_month) & (df['issue_date'].dt.day <= current_day) ] # ye hume ye dega ki issue date current month(june) ke equal ho or woh date aaj ke equal ya chhoti ho

no=a['id'].nunique()

print(no)



```

```python
total_amount=a['loan_amount'].sum()
print('total amount on current month till current day=',total_amount,'rupees')

# amount in millions

b=total_amount/1000000
c=round(b,2) # amount till 2 decimle

print('total amount on current month till current day=',c,'millions')
```

```python
good_loan=df.groupby('loan_status')['id'].agg('count')
good_loan1=df.groupby('loan_status').agg({
    'id':'count',
    'dti':'mean',
    'int_rate':'mean',
    'loan_amount':'sum',
    'total_payment':'sum'

})


good_loan1.rename(columns={'id':'Number of Application'},inplace=True)
display(good_loan1)

```

## Good Loan Analys..

```python
# no, of application
no_of_app=good_loan1.loc['Current','Number of Application']+ good_loan1.loc['Fully Paid','Number of Application']

print(no_of_app)

#funded amount

funded_amount=good_loan1.loc['Current','loan_amount']+ good_loan1.loc['Fully Paid','loan_amount']

print(funded_amount)


# recieved amount

recieved_amount=good_loan1.loc['Current','total_payment']+ good_loan1.loc['Fully Paid','total_payment']
print(recieved_amount)

# profit

print(recieved_amount-funded_amount)


```

```python
## Bad loan

no_of_app=good_loan1.loc['Charged Off','Number of Application']
print(no_of_app)

#funded amount

funded_amount=good_loan1.loc['Charged Off','loan_amount']

print(funded_amount)


# recieved amount

recieved_amount=good_loan1.loc['Charged Off','total_payment']
print(recieved_amount)

# profit

print(recieved_amount-funded_amount)
```
